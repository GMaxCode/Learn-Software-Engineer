AULA DE CLASSES

```python
class ContaCorrente:
    def __init__(self,titular,saldo_inicial):
        self.titular = titular
        self.saldo_inicial = saldo_inicial
    
    def depositar(self, valor):
        self.saldo_inicial += valor
    
    def sacar(self, valor):
        if valor > self.saldo_inicial:
            return'valor muito alto!!'
        self.saldo_inicial -= valor
        return f'você sacou: R${valor:.2f}'
        
        
    def mostrar(self):
        print(f"Saldo atual: R${self.saldo_inicial:.2f}")

conta  = ContaCorrente("Guilherme",900)
print(conta.depositar(90))
print(conta.sacar(33))
print(conta.mostrar())





        
```
