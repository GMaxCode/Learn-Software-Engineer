```python

class Tarefa:
    def __init__(self,titulo):
        self.titulo = titulo
        self.situacao = "A fazer"

    def concluir(self):
        self.situacao = "tarefa concluida"
    

matematica = Tarefa("tarefa de matematica")

print(matematica.situacao)
matematica.concluir()
print(matematica.situacao)
```

```python
class ContaBancaria:
    def __init__(self,titular,saldo=0.00):
        self.titular = titular
        self.saldo = saldo
        if self.saldo < 0:
            raise ValueError('valor muito baixo')

    def depositar(self,valor):
        if valor <=0:
            raise ValueError('valor menor que 0')
        self.saldo = valor + self.saldo

    def sacar(self,valor):
        if valor <=0:
            raise ValueError('o valor deve ser positivo')
        if valor > self.saldo:
            raise ValueError('saldo insuficiente no momento.')

        self.saldo -= valor

    def exibir_extrato(self):
        return f"Titular: {self.titular} | Saldo: R${self.saldo}"

try:

    next_conta = ContaBancaria('Guilherme',1000)
except ValueError as erro:
    print(f'erro: {erro} ')

try:

    next_conta.depositar(550)
except ValueError as erro:
    print(f'erro: {erro}')

try:

    next_conta.sacar(2000)

except ValueError as erro:
    print(f'erro: {erro}')



print(next_conta.saldo)
print(next_conta.titular)
print(next_conta.exibir_extrato())

    


        

```
