Praticando Classes

```python
class ContaBancaria:  #criação da classe conta bancaria
    def __init__(self, titular, saldoinicial):   # passando os parametros para os atributos da classe
        self.titular = titular
        self.saldo = saldoinicial    # aqui já começa com algum valor númerico int ou float 

    def depositar(self,valor):   # criação do método depositar 
        self.saldo = self.saldo + valor  # ao colocar um valor para o método (depositar) a variavel de objeto self.saldo já soma com esse valor.

        return self.saldo  # retorna o valor completo, deposito +  self.saldo


    def sacar(self, valor):    # criação do método sacar, onde o valor não pode ser nem maior que self.saldo nem menor que 0, precisa ser um valor menor que o saldo.

        if valor > self.saldo:
            raise ValueError(f'o valor {valor} é maior que o seu saldo na conta.')

        if valor <= 0:
            raise ValueError('valor menor ou = 0')

    
        self.saldo = self.saldo - valor 

        return self.saldo

 

conta = ContaBancaria('Guilherme',0)

print("--------- programa inicializado ----------")

deposito = conta.depositar(400)
valor_sacar = -5

print(f'você depositou: R$ {deposito:.2f}')

try:

    print(f'seu saldo é de: R$ {conta.sacar(valor_sacar):.2f}')

except ValueError as erro:
    print(f'erro {erro}')
    
print(conta.saldo)


    
```

```python
class Estoque:
    def __init__(self, produto: str, quantidade: int):
        self.produto = produto
        self.quantidade = quantidade

    def retirar(self, qtd: int):
        # 1. Valida ANTES de alterar o estado
        if qtd <= 0:
            raise ValueError("A quantidade deve ser positiva")

        # 2. Regra de negócio: não existe estoque negativo
        if qtd > self.quantidade:
            raise ValueError(
                f"Estoque insuficiente: pediu {qtd}, tem {self.quantidade}"
            )

        # 3. Só chega aqui se tudo estiver válido
        self.quantidade -= qtd




estoque = Estoque("Caneta", 10)

try:
    estoque.retirar(50)
except ValueError as erro:
    print(f"Não foi possível retirar: {erro}")

print("Fim do programa")
```
